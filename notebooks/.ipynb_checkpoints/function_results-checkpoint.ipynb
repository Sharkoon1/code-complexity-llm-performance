# Function-level LM-CC results

Metrics computed on the **gold patch's target function** (outermost top-level def/method), not the whole file. Question: does scoping remove the LOC size-confound?

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

from src.config import PATHS
import pandas as pd

metrics_df = pd.read_parquet(PATHS.FUNCTION_RESULT_DATASET)
labeled_df = pd.read_parquet(PATHS.LABELED_DATASET)
difficulty = pd.read_parquet(PATHS.TASK_DIFFICULTY)

merged = (
    metrics_df
    .merge(labeled_df[['instance_id', 'resolved']], on='instance_id', how='left')
    .merge(difficulty[['instance_id', 'resolution_rate']], on='instance_id', how='left')
)

clean = merged[
    merged['has_patched_function'].fillna(False)
    & merged['parsable'].fillna(False)
    & merged['lm_cc_score'].notna()
    & merged['resolution_rate'].notna()
].copy()

print(f'Total tasks:            {len(merged)}')
print(f'With patched function:  {int(merged["has_patched_function"].fillna(False).sum())}')
print(f'Excluded (no function): {int((~merged["has_patched_function"].fillna(False)).sum())}')
print(f'Usable (clean):         {len(clean)}')
```

## Headline diagnostic — is function LM-CC decoupled from function LOC?

```python
from scipy.stats import spearmanr
rho, _ = spearmanr(clean['lm_cc_score'], clean['loc'])
print(f'function-level rho(lm_cc_score, loc) = {rho:+.3f}   (whole-file was +0.982)')
print(f'function loc: median={int(clean["loc"].median())}, mean={clean["loc"].mean():.0f}  (whole-file median ~389)')
```

## Complexity vs. difficulty (resolution rate), controlling for function LOC

```python
from src.correlation import full_correlation_table

metrics_to_compare = [
    'cc_sum', 'halstead_volume', 'loc', 'lloc', 'nesting_max', 'n_branches',
    'lm_cc_total_comp', 'lm_cc_total_branch', 'lm_cc_max_comp', 'lm_cc_avg_comp',
    'lm_cc_max_branch', 'lm_cc_avg_branch', 'lm_cc_score',
]
available = [m for m in metrics_to_compare if m in clean.columns]
table = full_correlation_table(clean, available, score='resolution_rate', control='loc')
print('=== Function-level: correlation vs. resolution_rate (control = function LOC) ===')
print(table.to_string(index=False))
```

## Per-agent (function-level LM-CC) + difficulty scatter

```python
import matplotlib.pyplot as plt
from src.correlation import per_agent_correlation_table

all_predictions = pd.read_parquet(PATHS.ALL_PREDICTIONS)
long = all_predictions.merge(clean[['instance_id', 'lm_cc_score', 'loc']], on='instance_id', how='inner')
table1 = per_agent_correlation_table(long, metric='lm_cc_score', score='resolved', control='loc')
print(f'agents: {len(table1)}')
print(f'median sample_zero        = {table1["sample_zero"].median():+.3f}')
print(f'median sample_partial_loc = {table1["sample_partial_loc"].median():+.3f}')

plt.scatter(clean['resolution_rate'], clean['lm_cc_score'], alpha=0.4, s=15)
plt.xlabel('resolution rate'); plt.ylabel('function LM-CC')
plt.title('Function-level LM-CC vs task difficulty')
plt.show()
```
